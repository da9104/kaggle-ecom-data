```python
import matplotlib.pyplot as plt

top_states = df['customer_state'].value_counts().head(10)

plt.figure()
top_states.plot(kind='bar')
plt.title("Top 10 States by Number of Customers")
plt.xlabel("State")
plt.ylabel("Number of Customers")
plt.show()
```

```python
top_cities = df['customer_city'].value_counts().head(10)

plt.figure()
top_cities.plot(kind='bar')
plt.title("Top 10 Cities by Number of Customers")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.show()
```

```python
total_unique = df['customer_unique_id'].nunique()
total_ids =df['customer_id'].nunique()

repeat_rate = (total_ids - total_unique) / total_unique
repeat_rate
```

```python
top_states = df['customer_state'].value_counts().head(5)

plt.figure()
top_states.plot(kind='bar')
plt.title("Top 5 States")
plt.show()
```

## Key Insights

1. Most orders in the dataset were successfully delivered, indicating a reliable order fulfillment process.

2. São Paulo has the highest number of customers and orders, making it the most important market for the platform.

3. Credit cards are the most commonly used payment method among customers.

4. Delivery delays appear to negatively affect customer review scores.

5. Some product categories generate significantly more orders than others.
